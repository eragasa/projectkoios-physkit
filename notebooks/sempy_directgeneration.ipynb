```python
def excess_carrier_concentration_n(n_n: float, n_n0: float) -> float:
    """Return excess electron concentration above equilibrium."""
    return n_n - n_n0


def excess_carrier_concentration_p(p_n: float, p_n0: float) -> float:
    """Return excess hole concentration above equilibrium."""
    return p_n - p_n0
```
