# Transverse, Longitudinal, and Mixed Waves

We describe a one-dimensional material medium using a reference coordinate $X$.

The reference position of a material point is

$$
\mathbf{R}(X)=X\hat{\mathbf{e}}_x
$$

The current position is

$$
\mathbf{r}(X,t)=\mathbf{R}(X)+\mathbf{u}(X,t)
$$

where $\mathbf{u}(X,t)$ is the displacement field.

A wave propagating in the $X$-direction can have displacement parallel or perpendicular to the propagation direction.

## Transverse Wave

A transverse wave has displacement perpendicular to the direction of propagation:

$$
\mathbf{u}(X,t)
=
A\cos(kX-\omega t)\hat{\mathbf{e}}_y
$$

The wave propagates along $X$, while material points move in the $Y$-direction.

## Longitudinal Wave

A longitudinal wave has displacement parallel to the direction of propagation:

$$
\mathbf{u}(X,t)
=
A\cos(kX-\omega t)\hat{\mathbf{e}}_x
$$

The wave propagates along $X$, and material points also oscillate along $X$.

## Mixed Longitudinal-Transverse Wave

A mixed longitudinal-transverse wave has both components:

$$
\mathbf{u}(X,t)
=
A_L\cos(kX-\omega t)\hat{\mathbf{e}}_x
+
A_T\cos(kX-\omega t+\phi)\hat{\mathbf{e}}_y
$$

The longitudinal component moves material points along the propagation direction.  
The transverse component moves material points perpendicular to the propagation direction.

```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import HTML
from matplotlib.animation import FuncAnimation

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

N = 80
X = np.linspace(0.0, 4.0 * np.pi, N)

A = 0.35
k = 1.0
omega = 1.0

frames = 120
interval = 40


# ------------------------------------------------------------
# Shared plotting helper
# ------------------------------------------------------------


def setup_axis(ax, title):
    ax.set_title(title)
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlim(X.min() - 0.8, X.max() + 0.8)
    ax.set_ylim(-1.2, 1.2)
    ax.set_xlabel(r"propagation direction $X$")
    ax.set_ylabel(r"transverse direction $Y$")
    ax.axhline(0.0, linewidth=0.8)
    ax.grid(True, alpha=0.3)


# ------------------------------------------------------------
# 1. Transverse wave
# ------------------------------------------------------------


def animate_transverse_wave():
    fig, ax = plt.subplots(figsize=(8, 3))
    setup_axis(ax, "Transverse Wave")

    ax.text(
        0.02,
        0.90,
        r"$\mathbf{u}(X,t)=A\cos(kX-\omega t)\hat{\mathbf{e}}_y$",
        transform=ax.transAxes,
    )

    (line,) = ax.plot([], [], lw=2)
    points = ax.scatter([], [], s=24)

    def update(n):
        t = 2.0 * np.pi * n / frames

        x_current = X
        y_current = A * np.cos(k * X - omega * t)

        line.set_data(x_current, y_current)
        points.set_offsets(np.column_stack([x_current, y_current]))

        return line, points

    anim = FuncAnimation(
        fig,
        update,
        frames=frames,
        interval=interval,
        blit=True,
    )

    plt.close(fig)
    return HTML(anim.to_jshtml())


animate_transverse_wave()
```

```python
# ------------------------------------------------------------
# 2. Longitudinal wave
# ------------------------------------------------------------

def animate_longitudinal_wave():
    fig, ax = plt.subplots(figsize=(8, 3))
    setup_axis(ax, "Longitudinal Wave")

    ax.text(
        0.02,
        0.90,
        r"$\mathbf{u}(X,t)=A\cos(kX-\omega t)\hat{\mathbf{e}}_x$",
        transform=ax.transAxes,
    )

    equilibrium = ax.scatter(X, np.zeros_like(X), s=10, alpha=0.25)
    points = ax.scatter([], [], s=28)

    def update(n):
        t = 2.0 * np.pi * n / frames

        u_x = A * np.cos(k * X - omega * t)

        x_current = X + u_x
        y_current = np.zeros_like(X)

        points.set_offsets(np.column_stack([x_current, y_current]))

        return points, equilibrium

    anim = FuncAnimation(
        fig,
        update,
        frames=frames,
        interval=interval,
        blit=True,
    )

    plt.close(fig)
    return HTML(anim.to_jshtml())


animate_longitudinal_wave()
```

```python
# ------------------------------------------------------------
# 3. Mixed longitudinal-transverse wave
# ------------------------------------------------------------

def animate_mixed_wave():
    fig, ax = plt.subplots(figsize=(8, 3))
    setup_axis(ax, "Mixed Longitudinal-Transverse Wave")

    A_L = 0.28
    A_T = 0.35
    phi = np.pi / 2

    ax.text(
        0.02,
        0.87,
        r"$\mathbf{u}=A_L\cos(kX-\omega t)\hat{\mathbf{e}}_x"
        r"+A_T\cos(kX-\omega t+\phi)\hat{\mathbf{e}}_y$",
        transform=ax.transAxes,
    )

    equilibrium = ax.scatter(X, np.zeros_like(X), s=10, alpha=0.25)
    line, = ax.plot([], [], lw=2)
    points = ax.scatter([], [], s=24)

    def update(n):
        t = 2.0 * np.pi * n / frames

        u_x = A_L * np.cos(k * X - omega * t)
        u_y = A_T * np.cos(k * X - omega * t + phi)

        x_current = X + u_x
        y_current = u_y

        line.set_data(x_current, y_current)
        points.set_offsets(np.column_stack([x_current, y_current]))

        return line, points, equilibrium

    anim = FuncAnimation(
        fig,
        update,
        frames=frames,
        interval=interval,
        blit=True,
    )

    plt.close(fig)
    return HTML(anim.to_jshtml())


animate_mixed_wave()
```
